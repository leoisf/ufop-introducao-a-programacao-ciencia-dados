## Praticando lógica de programação com Python: controle de gastos mensais.

### Exercício 29
Você quer analisar seus gastos do mês e identificar quanto foi gasto apenas com despesas essenciais (valores múltiplos de 50 representam contas fixas como luz, água, aluguel).

Dada uma lista de gastos, some apenas os valores que são múltiplos de 50.

Entrada: gastos = [120, 50, 75, 200, 30, 150]
Saída esperada: 400

Lógica:
* Percorrer a lista
* Identificar múltiplos de 50 (% 50 == 0)
* Somar apenas esses valores



```python
gastos = [120, 50, 75, 200, 30, 150]
total_essenciais = 0

for gasto in gastos:
    if gasto % 50 == 0:
        total_essenciais += gasto

print("Total de despesas essenciais:", total_essenciais)
```

### Exercício 30
Você trabalha em um estoque e precisa saber quantos produtos estão com quantidade par (facilita divisão em caixas).

Conte quantos produtos têm quantidade par.

Entrada: estoque = [15, 20, 33, 40, 18, 7]
Saída esperada: 3

Lógica:
* Percorrer a lista
* Verificar se é par (%2==0)
* Contar



```python
estoque = [15, 20, 33, 40, 18, 7]
contagem_pares = 0

for quantidade in estoque:
    if quantidade % 2 == 0:
        contagem_pares += 1

print("Produtos com quantidade par:", contagem_pares)
```

# Exercício 31 - Maior venda do dia sem usar max()


```python
vendas = [150, 320, 90, 500, 210]

maior = vendas[0]  # começa com o primeiro valor

for venda in vendas:
    if venda > maior:
        maior = venda  # atualiza quando encontrar maior

print("Maior venda do dia: R$", maior)
```
