## 1. Definição do Problema
**Dataset escolhido:** *Tips* (Gorjetas em um restaurante).
**Pergunta/Problema:** Existe diferença significativa no valor das gorjetas entre o almoço e o jantar? Como isso pode impactar a escala de funcionários?

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Carregando um dataset clássico e simples do Seaborn
df = sns.load_dataset('tips')
print("Dataset 'Tips' carregado com sucesso!")
```

## 2. Exploração dos Dados
Vamos utilizar as funções `head()`, `info()` e `describe()` para entender a estrutura dos dados.

## 3. Visualização
Analisando a distribuição das gorjetas por período do dia.

```python
plt.figure(figsize=(8, 5))
sns.boxplot(x='time', y='tip', data=df, palette='Set2')
plt.title('Distribuição de Gorjetas: Almoço vs Jantar')
plt.xlabel('Período')
plt.ylabel('Valor da Gorjeta ($)')
plt.show()
```

## 4. Padrões e Insights
1. **Padrão:** O gráfico de boxplot mostra que o valor mediano das gorjetas é ligeiramente superior no jantar.
2. **Padrão:** Há uma maior presença de *outliers* (gorjetas excepcionalmente altas) durante o jantar.

## 5. Decisão Baseada em Dados
**Proposta:** Como o volume de gorjetas e os valores absolutos são maiores no jantar, a gerência deve **alocar os garçons mais experientes para o turno da noite**, visando manter o alto nível de serviço onde o potencial de ganho e ticket médio é maior.
