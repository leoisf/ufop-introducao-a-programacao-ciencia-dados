## 1. Lista de números pares

Crie uma lista com os números de 1 a 50 usando `range()`.

Depois:
- gere uma nova lista contendo apenas os números pares;
- gere outra lista contendo o quadrado de cada número.

````python
numeros = list(range(1, 51))
print("Lista de 1 a 50:")
print(numeros)

pares = [n for n in numeros if n % 2 == 0]
print("\nNúmeros pares:")
print(pares)

quadrados = [n ** 2 for n in numeros]
print("\nQuadrado de cada número:")
print(quadrados)
````

## 2. Manipulação de listas

Dada a lista:

```python
nomes = ["Ana", "Carlos", "Maria", "Pedro", "João"]
```

Faça:
- adicione um novo nome ao final;
- insira um nome na posição 2;
- remova um nome;
- ordene a lista em ordem alfabética;
- exiba a lista invertida.

````python
nomes = ["Ana", "Carlos", "Maria", "Pedro", "João"]
print("Lista original:", nomes)

nomes.append("Lucas")
print("\nApós append('Lucas'):", nomes)

nomes.insert(2, "Beatriz")
print("Após insert(2, 'Beatriz'):", nomes)

nomes.remove("Pedro")
print("Após remove('Pedro'):", nomes)

nomes.sort()
print("\nLista ordenada alfabeticamente:", nomes)

nomes_invertidos = nomes[::-1]
print("Lista invertida:", nomes_invertidos)
````

## 4. Dicionário de alunos

Crie um dicionário contendo:
- nome do aluno;
- idade;
- curso;
- notas (lista com 3 notas).

Depois:
- calcule a média das notas;
- adicione a chave `"média"` ao dicionário;
- exiba todas as chaves e valores usando `items()`.

````python
aluno = {
    "nome": "Leonardo",
    "idade": 22,
    "curso": "Ciência de Dados",
    "notas": [7.5, 8.0, 9.0]
}

media = sum(aluno["notas"]) / len(aluno["notas"])

aluno["média"] = round(media, 2)

print("Dados do aluno:")
for chave, valor in aluno.items():
    print(f"  {chave}: {valor}")
````

## 5. Contagem de palavras

Peça ao usuário uma frase.

Depois:
- separe as palavras;
- conte quantas vezes cada palavra aparece;
- armazene o resultado em um dicionário.

````python
frase = "o gato subiu no telhado e o cachorro ficou no quintal e o gato desceu"
print(f"Frase: \"{frase}\"")

palavras = frase.lower().split()

contagem = {}
for palavra in palavras:
    contagem[palavra] = contagem.get(palavra, 0) + 1

print("\nContagem de palavras:")
for palavra, qtd in sorted(contagem.items(), key=lambda x: x[1], reverse=True):
    print(f"  '{palavra}': {qtd} vez(es)")
````

## 6. Operações com sets

Crie dois conjuntos:

```python
a = {1, 2, 3, 4, 5}
b = {4, 5, 6, 7, 8}
```

Faça:
- união;
- interseção;
- diferença;
- diferença simétrica;
- verifique se `{1, 2}` é subconjunto de `a`.

````python
a = {1, 2, 3, 4, 5}
b = {4, 5, 6, 7, 8}

print(f"Conjunto a: {a}")
print(f"Conjunto b: {b}")

uniao = a | b
print(f"\nUnião (a | b):                {uniao}")

intersecao = a & b
print(f"Interseção (a & b):            {intersecao}")

diferenca = a - b
print(f"Diferença (a - b):             {diferenca}")

dif_simetrica = a ^ b
print(f"Diferença simétrica (a ^ b):   {dif_simetrica}")

subconjunto = {1, 2}.issubset(a)
print(f"\n{{1, 2}} é subconjunto de a? {subconjunto}")
````

## 7. Funções

Crie uma função chamada `calcular_media()` que:
- receba uma lista de notas;
- retorne a média.

Depois:
- crie outra função chamada `situacao()` que receba a média e retorne:
  - `"Aprovado"` se média ≥ 60;
  - `"Recuperação"` se média entre 40 e 59;
  - `"Reprovado"` caso contrário.

````python
def calcular_media(notas):
    """Recebe uma lista de notas e retorna a média."""
    if not notas:
        return 0
    return sum(notas) / len(notas)


def situacao(media):
    """Recebe a média e retorna a situação do aluno."""
    if media >= 60:
        return "Aprovado"
    elif media >= 40:
        return "Recuperação"
    else:
        return "Reprovado"


exemplos = [
    [80, 75, 90],
    [50, 45, 55],
    [20, 30, 25]
]

for notas in exemplos:
    media = calcular_media(notas)
    resultado = situacao(media)
    print(f"Notas: {notas} → Média: {media:.1f} → Situação: {resultado}")
````

## 8. Lambda e comprehensions

Dada a lista:

```python
numeros = [1, 2, 3, 4, 5, 6, 7, 8, 9]
```

Faça:
- use `lambda` com `map()` para gerar o dobro dos números;
- use list comprehension para gerar apenas números ímpares;
- use `filter()` para gerar números maiores que 5.

````python
numeros = [1, 2, 3, 4, 5, 6, 7, 8, 9]
print("Lista original:", numeros)

dobros = list(map(lambda x: x * 2, numeros))
print("\nDobro de cada número (map + lambda):", dobros)

impares = [n for n in numeros if n % 2 != 0]
print("Números ímpares (list comprehension):", impares)

maiores_que_5 = list(filter(lambda x: x > 5, numeros))
print("Números maiores que 5 (filter + lambda):", maiores_que_5)
````

## 9. Leitura de arquivos

Crie um arquivo chamado `texto.txt` contendo algumas frases.

Depois:
- abra o arquivo;
- leia todas as linhas;
- conte:
  - quantidade de linhas;
  - quantidade de palavras;
  - quantidade de caracteres.

````python
import os

# Criando o arquivo texto.txt com algumas frases
conteudo = """Python é uma linguagem de programação poderosa.
Ciência de dados usa Python para análise e visualização.
Listas, dicionários e sets são estruturas fundamentais.
Funções tornam o código mais organizado e reutilizável.
Aprender programação abre muitas portas profissionais.
"""

caminho = "texto.txt"

with open(caminho, "w", encoding="utf-8") as f:
    f.write(conteudo)

print(f"Arquivo '{caminho}' criado com sucesso!")

with open(caminho, "r", encoding="utf-8") as f:
    linhas = f.readlines()

linhas_validas = [l for l in linhas if l.strip()]

total_linhas = len(linhas_validas)
total_palavras = sum(len(l.split()) for l in linhas_validas)
total_caracteres = sum(len(l.strip()) for l in linhas_validas)

print("\nConteúdo do arquivo:")
for i, linha in enumerate(linhas_validas, 1):
    print(f"  Linha {i}: {linha.strip()}")

print(f"\nEstatísticas:")
print(f"  Quantidade de linhas:     {total_linhas}")
print(f"  Quantidade de palavras:   {total_palavras}")
print(f"  Quantidade de caracteres: {total_caracteres}")
````

## 10. Sistema simples de cadastro

Crie um programa que:
- permita cadastrar nomes de alunos;
- armazene os nomes em um arquivo `.txt`;
- permita listar todos os alunos cadastrados;
- trate nomes duplicados usando `set`.

````python
import os

ARQUIVO_ALUNOS = "alunos.txt"


def carregar_alunos():
    """Carrega os alunos do arquivo e retorna um set (evita duplicatas)."""
    if not os.path.exists(ARQUIVO_ALUNOS):
        return set()
    with open(ARQUIVO_ALUNOS, "r", encoding="utf-8") as f:
        return set(linha.strip() for linha in f if linha.strip())


def salvar_alunos(alunos):
    """Salva o set de alunos no arquivo."""
    with open(ARQUIVO_ALUNOS, "w", encoding="utf-8") as f:
        for nome in sorted(alunos):
            f.write(nome + "\n")


def cadastrar_aluno(nome, alunos):
    """Cadastra um aluno, ignorando duplicatas."""
    nome = nome.strip().title()
    if nome in alunos:
        print(f"  ⚠ '{nome}' já está cadastrado (duplicata ignorada).")
    else:
        alunos.add(nome)
        salvar_alunos(alunos)
        print(f"  ✔ '{nome}' cadastrado com sucesso!")


def listar_alunos(alunos):
    """Lista todos os alunos cadastrados."""
    if not alunos:
        print("  Nenhum aluno cadastrado.")
    else:
        print(f"  Total de alunos: {len(alunos)}")
        for i, nome in enumerate(sorted(alunos), 1):
            print(f"  {i}. {nome}")


alunos = carregar_alunos()

nomes_para_cadastrar = ["Ana Silva", "Carlos Souza", "Maria Oliveira",
                        "Pedro Lima", "Ana Silva", "João Santos", "carlos souza"]

print("=== Cadastro de Alunos ===")
for nome in nomes_para_cadastrar:
    cadastrar_aluno(nome, alunos)

print("\n=== Lista de Alunos Cadastrados ===")
listar_alunos(alunos)

print(f"\nArquivo '{ARQUIVO_ALUNOS}' salvo com sucesso!")
````
